```python
import pandas as pd
from transformers import pipeline
```

```python
modello = pipeline(
    "token-classification",
    model=r"C:\AI\models\rizzo-pii-0.3B",
    device="cpu",
    aggregation_strategy="simple",
    #local_files_only=True
)
```

```python
testo = """
Mario Rossi, nato a Roma (RM) il 12 maggio 1985,
lavora presso Alfa S.r.l.
La sua email è mario.rossi@example.com.
"""

risultati = modello(testo)

for entita in risultati:
    print(
        entita["word"],
        entita["entity_group"],
        round(entita["score"], 3)
    )
```
